```python
import gymnasium as gym
import numpy as np
```

```python
# Initialise the environment

reward_goal, reward_hole, reward_frozen = 1.0,0.0,0.0

env = gym.make("FrozenLake-v1", 
               map_name="4x4",
               render_mode="ansi",
               is_slippery=False,
               success_rate=2.0/3.0,
               reward_schedule = (reward_goal, reward_hole, reward_frozen))
```

S -> Start

F -> Frozen

H -> Hole

G -> Goal

```python
# 10 random steps

# Reset the environment to generate the first observation
observation, info = env.reset(seed=42)

print(env.render())

for _ in range(10):
    # this is where you would insert your policy
    action = env.action_space.sample()

    # step (transition) through the environment with the action
    # receiving the next observation, reward and if the episode has terminated or truncated
    observation, reward, terminated, truncated, info = env.step(action)
    
    # If the episode has ended then we can reset to start a new episode
    if terminated or truncated:
        observation, info = env.reset()

    print(env.render())

env.close()
```

```python
# Initialization
observation, info = env.reset(seed = 12)

print('Observation: ', observation)
```

```python
# Inspect the MDP parts

# action space
print('Action space:', env.action_space)

# observation space
print('Observation space:', env.observation_space)

# reward function
print('Rewards:' )
print(f'\t reach goal -> {reward_goal}')
print(f'\t reach hole -> {reward_hole}')
print(f'\t reach frozen -> {reward_frozen}')
```

Decodificar Ações:

0: Mover para esquerda

1: Mover para baixo

2: Mover para direita

3: Mover para cima

```python
# Análise do output de env.step(action):

def obs_to_grid(observation, grid_side_len = 4):
    col = observation % grid_side_len
    row = int((observation - col)/grid_side_len)
    return row, col

row, col =  obs_to_grid(observation)

print(f'Nova observação é: {observation}')
print(f'\t Posição no grid é: {row} X {col}')
print(env.render())
```

```python
# Reward é o escalar associado à transição S -> F
print(f'Reward obtida: {reward}')
# Terminated é um booleano que diz se o environment terminou nessa transição:
print(f'Terminated? {terminated}')
# Truncated é um booleano que diz o environment foi interrompido (e.g.: limite de steps):
print(f'Truncated? {terminated}')
# info é um dicionário com informações extras para quem desenhou ou vai operar no environment.
#   Nesse caso, informa a probabilidade da transição que ocorreu (sempre 1.0, caso determinístico)
print(f'info = {info} ')
```
